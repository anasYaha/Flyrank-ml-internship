## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:** Given a content page's observable search and engagement signals over the last 90 days, which pages should a content reviewer look at first — and why?

- **Unit of analysis:** one content page (content_id) within one client (client_id), evaluated over a 90-day observation window.
- **Decision improved:** which pages a content/SEO reviewer spends their limited review time on this cycle, out of potentially thousands of pages per client.
- **Who acts on it:** a human content reviewer or SEO strategist at FlyRank (or at a client), who then decides on an action such as refresh, expand, protect, prune, or monitor.
- **Cost of a wrong recommendation:**
  - False positive (page flagged but didn't need it): wasted reviewer time, less time for pages that actually needed attention.
  - False negative (a genuinely declining, high-value page not flagged): traffic keeps leaking unnoticed until the next review cycle — the costlier mistake.
- **Why data/ML can help at all:** with hundreds of thousands of content items and no way for a human to check every page every cycle, a ranked, evidence-backed queue turns an unmanageable full audit into a short, prioritized shortlist.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

csv_url = "https://raw.githubusercontent.com/anasYaha/Flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(csv_url)

# Apply the same filters the starter pipeline uses, so numbers are comparable
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

total_pages = len(df)
pct_declining = (df["trend_direction"] == "down").mean() * 100

stale_visible = df[
    (df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)
]
declining_with_demand = df[
    (df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)
]

print(f"Total visible, mature pages in starter slice: {total_pages}")
print(f"Share currently trending down: {pct_declining:.1f}%")
print(f"Stale + visible pages (>=180d since update, >=500 impressions): {len(stale_visible)}")
print(f"Declining pages that still have real demand (>=100 impressions): {len(declining_with_demand)}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What I can say:**
- Which pages show observed signs of decline, staleness, low CTR relative to their position, or weak engagement, based on the last 90 days of data.

- That refreshing a page caused a recovery — that requires an actual experiment, not observational data.
